Q4) a) Explain the primary purpose of the dataset splits

- Training Set

Used to teach the model.

- Validation Set

Used to tune hyperparameters and choose the best model.

- Test Set

Used only once at the end to evaluate performance on unseen data.

b) What are advantages of evaluating a machine learning model using a separate
test dataset?

Gives an unbiased evaluation

Measures how well the model generalises

Helps detect overfitting

c)  Suppose a student trains and evaluates a model using the same dataset and
obtains an accuracy of 99%. Would this result necessarily indicate that the model
performs well on unseen data? Justify your answer.

- No.

Using the same data for training and testing does not necessarily mean the model performs well on unseen data. The model may have memorised the training data, leading to an overly optimistic accuracy. A separate test dataset is required to measure true generalisation performance.
